```python
import torch
import numpy as np

print("COMP395 Workstation GPU Verification")
print("=" * 45)

print("PyTorch:", torch.__version__)
print("NumPy:", np.__version__)
print("CUDA:", torch.version.cuda)
print("CUDA Available:", torch.cuda.is_available())
print("GPU Count:", torch.cuda.device_count())

for i in range(torch.cuda.device_count()):
    device = torch.device(f"cuda:{i}")

    a = torch.ones((512, 512), device=device)
    result = a @ a
    torch.cuda.synchronize(device)

    value = result[0, 0].item()

    print(f"\nGPU {i}: {torch.cuda.get_device_name(i)}")
    print(f"VRAM: {torch.cuda.get_device_properties(i).total_memory / 1024**3:.2f} GiB")
    print(f"Result: {value}")
    print(f"Status: {'PASS' if value == 512.0 else 'FAIL'}")
```
